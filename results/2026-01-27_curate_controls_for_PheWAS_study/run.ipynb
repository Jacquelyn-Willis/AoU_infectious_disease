```python
def import_map_files():
   
    
    
    phe_icd= pd.read_csv(f"{data}/expanded_phecode.csv")
    
    phe = pd.read_csv(f"{data}/phecode_info.csv")
    
    phex = pd.read_csv(f"{data}/phecodex_info.csv")
    
    phex = phex.rename({"phecodex": "phecode"}, axis="columns")
    
    phex_icd = pd.read_csv(f"{data}/updated_phecodex_map.csv")
    
    
    return phe, phe_icd, phex, phex_icd
    
```

```python
def merge_phecodes_with_phenotype(phe, phe_icd):
    
    final = pd.merge(phe, phe_icd, how = "inner")
    
    return final
```

```python
def filter_map_for_phex_cohorts(phemap, phemapx): 
    
    
    subset = phemap[phemap["description"].isin(["viral warts & hpv", "hiv infection, symptomatic", "varicella infection", "herpes simplex", "viral infection", "viral hepatitis c", "viral hepatitis"])]
    
    subsetx = phemapx[phemapx["description"].isin(["sars-cov-2*", "viral pneumonia", "influenza virus", "hepatitis b"])]
    
    print(subset["description"].unique())
    print(subsetx["description"].unique())
   
    
    return subset, subsetx
```

```python
def get_data_pkl(path, filename):

    # …later, in any notebook in the same workspace…
    out_file = Path(f'{path}/{filename}')

    # Reload:
    with open(out_file, 'rb') as f:
        cohort_data_dict = pickle.load(f)

    #print("Reloaded keys:", list(cohort_data_dict.keys()))
    
    return cohort_data_dict
```

```python
def subset_my_cohort_by_phecode(cohort_df):
    
    
    #phecode

    hpv = cohort_df[cohort_df["standard_concept_name"].isin(["Verruca vulgaris", "Verruca plantaris", "Condyloma acuminatum of the anogenital region", "Human papilloma virus infection", "Genital warts"])]
    hiv = cohort_df[cohort_df["standard_concept_name"].isin(["Human immunodeficiency virus infection", "Asymptomatic human immunodeficiency virus infection"])]
    vzv = cohort_df[cohort_df["standard_concept_name"] == "Varicella"]
    viral_inf = cohort_df[cohort_df["standard_concept_name"] == "Viral disease"]
    hep_c = cohort_df[cohort_df["standard_concept_name"].isin(["Viral hepatitis C","Chronic hepatitis C", "Acute hepatitis C", "Chronic viral hepatitis C with hepatic coma"])]
    viral_hep = cohort_df[cohort_df["standard_concept_name"].isin(["Viral hepatitis without hepatic coma", "Acute viral hepatitis", "Viral hepatitis", "Viral hepatitis in mother complicating pregnancy", "Chronic viral hepatitis"])]
    hsv = cohort_df[cohort_df["standard_concept_name"].isin(["Herpes simplex","Herpetic vulvovaginitis", "Anogenital herpesviral infection", "Herpes simplex keratitis", "Herpes labialis	Herpes simplex disciform keratitis", "Herpes simplex with complication", "Ophthalmic herpes simplex", "Recurrent herpes simplex labialis", "Herpetic infection of penis", "Herpetic ulceration of vulva", "Herpesviral infection of perianal skin and rectum"])]
    
    #phecodeX
    
    cov = cohort_df[cohort_df["standard_concept_name"].isin(["COVID-19", "Pneumonia caused by SARS-CoV-2", "Disease due to Coronaviridae", "Coronavirus infection"])]
    viral_pneu = cohort_df[cohort_df["standard_concept_name"] == "Viral pneumonia"]
    flu = cohort_df[cohort_df["standard_concept_name"].isin(["Upper respiratory tract infection due to Influenza", "Influenza", "Pneumonia and influenza", "Influenza caused by seasonal influenza virus", "Influenzal acute upper respiratory infection", "Influenza with non-respiratory manifestation"])]
   
    hep_b = cohort_df[cohort_df["standard_concept_name"].isin(["Viral hepatitis B without hepatic coma", "Chronic viral hepatitis B without delta-agent", "Acute type B viral hepatitis"])]

    return hpv, hiv, vzv, viral_inf, hep_c, viral_hep, hsv, cov, viral_pneu, flu, hep_b
    
    
    
```
