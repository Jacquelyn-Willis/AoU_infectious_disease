```python
merged = pd.read_csv(f"{scratch}/matched_covid_19_omicron_white_y_matchit_cohort_merged.tsv", sep="\t")

pc_cols = [c for c in merged.columns if c.upper().startswith("PC")]
missing_pc_mask = merged[pc_cols].isna().all(axis=1)

print("Total individuals:", len(merged))
print("Missing all PCs:", missing_pc_mask.sum())

# Inspect a few IDs that are missing PCs
merged.loc[missing_pc_mask, ["FID", "IID"]].head()
```

```python
pcs = pd.read_csv(f"{results}/wrangled_ancestry_pcs.csv")  # same path you used in batch_process
pcs_ids = set(pcs["research_id"].astype(str).str.strip())

missing_ids = merged.loc[missing_pc_mask, "IID"].astype("string")

# How many of these IIDs are actually present in the PC file?
print("Missing IDs that DO appear in PCs:",
      missing_ids.isin(pcs_ids).sum())

print("Missing IDs that do NOT appear in PCs:",
      (~missing_ids.isin(pcs_ids)).sum())
```
