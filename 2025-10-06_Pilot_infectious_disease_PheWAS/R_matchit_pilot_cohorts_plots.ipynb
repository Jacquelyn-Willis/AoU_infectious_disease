```r
# Make sure ggplot2 is loaded
# library(ggplot2)

plot_basic_distributions <- function(your_dataframe) {
  
  # Ensure ggplot2 is available
  if (!requireNamespace("ggplot2", quietly = TRUE)) {
    stop("Package 'ggplot2' is needed for this function. Please install it.")
  }
  
  # --- 1. Plot for sex_at_birth (Bar Chart) ---
  sex_plot <- ggplot2::ggplot(your_dataframe, ggplot2::aes(x = sex_at_birth, fill = sex_at_birth)) +
    ggplot2::geom_bar() +
    ggplot2::labs(
      title = "Distribution of Sex at Birth",
      x = "Sex at Birth",
      y = "Count"
    ) +
    ggplot2::theme_bw() +
    ggplot2::theme(legend.position = "none")
  
  # --- 2. Plot for updated_race (Bar Chart) ---
  race_plot <- ggplot2::ggplot(your_dataframe, ggplot2::aes(x = updated_race, fill = updated_race)) +
    ggplot2::geom_bar() +
    ggplot2::labs(
      title = "Distribution of Race / Ethnicity",
      x = "Race / Ethnicity",
      y = "Count"
    ) +
    ggplot2::theme_bw() +
    ggplot2::theme(
      legend.position = "none",
      axis.text.x = ggplot2::element_text(angle = 45, hjust = 1) 
    )
  
  # --- 3. Plot for age (Histogram) ---
  age_plot <- ggplot2::ggplot(your_dataframe, ggplot2::aes(x = age)) +
    ggplot2::geom_histogram(bins = 30, fill = "steelblue", color = "black") +
    ggplot2::labs(
      title = "Age Distribution",
      x = "Age (Years)",
      y = "Frequency (Count)"
    ) +
    ggplot2::theme_bw()
  
  # --- 4. Print all plots ---
  # In R, you must explicitly print() ggplot objects when inside a function or loop
  print(sex_plot)
  print(race_plot)
  print(age_plot)
}
```
