```r
#Merge case and controls in subset dataframes for each cohort
```

```r
#POLIO
# ----------------------------
# 0) Start from your two data frames
# ----------------------------
cases    <- polio      # e.g., cases_df
controls <- polio_controls   # e.g., controls_df

# ----------------------------
# 1) Harmonize column names (EDIT the <> pieces)
# ----------------------------
cases <- cases |>
  mutate(case = 1L)

controls <- controls |>
  mutate(case = 0L)

# ----------------------------
# 2) Bind, clean types, drop missing
# ----------------------------
df <- bind_rows(cases, controls)

df <- df |>
  mutate(
    sex = factor(sex_at_birth),
    ancestry = factor(updated_race),
    age = as.numeric(age)
  ) 


# Quick sanity checks
table(df$case)
table(df$sex, df$case)
table(df$ancestry, df$case)
summary(df$age)
```

```r
#covid
# ----------------------------
# 0) Start from your two data frames
# ----------------------------
cases2    <- covid      # e.g., cases_df
controls2 <- covid_controls   # e.g., controls_df

# ----------------------------
# 1) Harmonize column names (EDIT the <> pieces)
# ----------------------------
cases2 <- cases2 |>
  mutate(case = 1L)

controls2 <- controls2 |>
  mutate(case = 0L)

# ----------------------------
# 2) Bind, clean types, drop missing
# ----------------------------
df2 <- bind_rows(cases2, controls2)

df_cov <- df2 |>
  mutate(
    sex = factor(sex_at_birth),
    ancestry = factor(updated_race),
    age = as.numeric(age)
  ) 


# Quick sanity checks
table(df_cov$case)
table(df_cov$sex, df_cov$case)
table(df_cov$ancestry, df_cov$case)
summary(df_cov$age)
```

```r
#run matchit on merged wrangled dataframes
```

```r
# Assumes 'my_data' has columns: 'case_status' (1/0), 'age', 'sex', 'ancestry'
# We'll match 5 controls to 1 case (ratio = 5), which is good for power in a PheWAS

m.out <- matchit(case ~ age + sex + ancestry,
                 data = df,
                 method = "nearest",    # Use nearest neighbor...
                 exact = c("sex", "ancestry"), # ...but ONLY within exact groups of sex and ancestry
                 ratio = 10,             # Request 10 controls per case
                 distance = "mahalanobis") # Match on 'age' (as it's the only one left)


summary(m.out)          # balance diagnostics
love.plot(m.out)        # optional: pretty balance plot

# ----------------------------
# 4) Get matched/weighted cohorts
# ----------------------------
matched <- match.data(m.out)  # includes weights and subclass

```

```r
write.csv(matched, "cohort_data/matched_polio_case_control_cohorts.csv", row.names = FALSE)
```

```r
# Assumes 'my_data' has columns: 'case_status' (1/0), 'age', 'sex', 'ancestry'
# We'll match 5 controls to 1 case (ratio = 5), which is good for power in a PheWAS

m2.out <- matchit(case ~ age + sex + ancestry,
                 data = df_cov,
                 method = "nearest",    # Use nearest neighbor...
                 exact = c("sex", "ancestry"), # ...but ONLY within exact groups of sex and ancestry
                 ratio = 5,             # Request 10 controls per case
                 distance = "mahalanobis") # Match on 'age' (as it's the only one left)


summary(m2.out)          # balance diagnostics
love.plot(m2.out)        # optional: pretty balance plot

# ----------------------------
# 4) Get matched/weighted cohorts
# ----------------------------
matched2 <- match.data(m2.out)  # includes weights and subclass
```
